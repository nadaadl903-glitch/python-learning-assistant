This notebook builds and evaluates a Retrieval-Augmented Generation (RAG) pipeline for a Python Learning Assistant.

The pipeline:
1. Loads and inspects the source documents.
2. Splits documents into chunks.
3. Generates embeddings.
4. Stores embeddings in ChromaDB.
5. Retrieves relevant chunks.
6. Builds a grounded prompt.
7. Generates answers using a local Ollama LLM.
8. Evaluates the pipeline on 10 questions.

```python
from pathlib import Path
from langchain_community.document_loaders import DirectoryLoader, TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_community.llms import Ollama

DATA_PATH = Path("../data")
VECTOR_STORE_PATH = Path("../chroma_db")
print("Data folder:", DATA_PATH.resolve())
print("Vector store:", VECTOR_STORE_PATH.resolve())
```

The project uses six .txt source documents covering Python basics, variables and data types, functions, object-oriented programming, exception handling, and file handling.

The documents are text files, so no OCR is required.

```python
loader = DirectoryLoader(
    str(DATA_PATH),
    glob="*.txt",
    loader_cls=TextLoader
)

documents = loader.load()

print("Number of documents:", len(documents))
print("\nLoaded files:")

for document in documents:
    print("-", document.metadata.get("source"))

print("\nAll documents loaded successfully.")
```
